```python
import random
import matplotlib.pyplot as plt

# 1. DADOS DO PROBLEMA
NOME_ITENS = ["Celular", "Notebook", "Livro", "Casaco", "Tênis", "Fone"]
PESOS = [1, 5, 2, 3, 2, 1]
VALORES = [400, 500, 30, 150, 200, 100]

CAPACIDADE_MOCHILA = 8  # Peso máximo permitido
NUM_GENES = 6           # Temos 6 itens, então 6 genes (0 ou 1)
TAM_POPULACAO = 10       # Definindo o tamanho da população
NUM_GERACOES = 100      # Número de gerações para o algoritmo evoluir
CHANCE_MUTACAO = 0.1    # Probabilidade de ocorrer mutação (10%)

# Função para criar indivíduo aleatório
def criar_individuo():
    return [random.randint(0, 1) for _ in range(NUM_GENES)]

# Função de fitness
def calcular_fitness(individuo):
    peso_total = 0
    valor_total = 0

    # Percorre cada gene
    for i in range(NUM_GENES):
        gene = individuo[i]
        if gene == 1:  # item selecionado
            peso_total += PESOS[i]
            valor_total += VALORES[i]

    # Penalidade se passar do limite
    if peso_total > CAPACIDADE_MOCHILA:
        return 0
    else:
        return valor_total

# Criar população inicial
populacao = []
for i in range(TAM_POPULACAO):
    novo_ind = criar_individuo()
    populacao.append(novo_ind)
```

```python
def selecao_roleta(populacao):
 # Sorteia dois índices aleatórios da população
 indice1 = random.randint(0, len(populacao) - 1)
 indice2 = random.randint(0, len(populacao) - 1)
 pai1 = populacao[indice1]
 pai2 = populacao[indice2]
 # Compara quem tem o melhor fitness
 fitness1 = calcular_fitness(pai1)
 fitness2 = calcular_fitness(pai2)
 if fitness1 > fitness2:
  return pai1
 else:
  return pai2
```

```python
def crossover(pai1, pai2):
 """Corta ao meio e troca as partes."""
 # Define o ponto de corte no meio exato (pode ser aleatório também)
 ponto_corte = 3
 filho1 = []
 filho2 = []
 # Construindo Filho 1 (Metade Pai1 + Metade Pai2)
 # Parte 1
 for i in range(0, ponto_corte):
  filho1.append(pai1[i])
 # Parte 2
 for i in range(ponto_corte, NUM_GENES):
  filho1.append(pai2[i])
 # Construindo Filho 2 (Metade Pai2 + Metade Pai1)
 # Parte 1
 for i in range(0, ponto_corte):
  filho2.append(pai2[i])
 # Parte 2
 for i in range(ponto_corte, NUM_GENES):
  filho2.append(pai1[i])
 return filho1, filho2
```

```python
def mutacao(individuo):
 """Percorre a lista e pode mudar o bit."""
 for i in range(NUM_GENES):
 # Joga o dado para ver se muda
  chance = random.random() # Gera numero entre 0.0 e 1.0
 if chance < CHANCE_MUTACAO:
 # Se era 0 vira 1, se era 1 vira 0
  if individuo[i] == 0:
   individuo[i] = 1
 else:
  individuo[i] = 0
 # Não precisa retornar, pois altera a lista original
```

```python
# EXECUÇÃO PRINCIPAL
# 1. Criar População Inicial
populacao = []
for i in range(TAM_POPULACAO):
 novo_ind = criar_individuo()
 populacao.append(novo_ind)
```

```python
lista_melhores_fitness = [] # Para o gráfico
 # Loop das Gerações
for geracao in range (NUM_GERACOES):
 # --- Avaliação e Estatísticas ---
  melhor_fitness_da_geracao = 0
 # Acha o melhor dessa geração atual
  for ind in populacao:
    fit = calcular_fitness(ind)
    if fit > melhor_fitness_da_geracao:
      melhor_fitness_da_geracao = fit
  lista_melhores_fitness.append(melhor_fitness_da_geracao)
 # print(f"Geração {geracao}: Melhor Fitness = {melhor_fitness_da_geracao}")
 # --- Reprodução ---
  nova_populacao = []
 # Criamos novos indivíduos até encher a nova população
  while len(nova_populacao) < TAM_POPULACAO:
 # 1. Selecionar Pais
    pai_a = selecao_roleta(populacao)
    pai_b = selecao_roleta(populacao)
 # 2. Cruzar (Ter filhos)
    filho_a, filho_b = crossover(pai_a, pai_b)
 # 3. Mutação (Chance de mudar)
    mutacao(filho_a)
    mutacao(filho_b)
 # 4. Adicionar na nova geração
    nova_populacao.append(filho_a)
    nova_populacao.append(filho_b)
 # Substitui a geração antiga pela nova
    populacao = nova_populacao
```

```python
# RESULTADO FINAL
print("--- Fim da Evolução ---")
print(f"Melhor valor encontrado na última geração: R$ {lista_melhores_fitness[-1]}")
# Mostrar gráfico
plt.plot(lista_melhores_fitness)
plt.title("Evolução do Algoritmo")
plt.xlabel("Geração")
plt.ylabel("Valor (Fitness)")
plt.grid(True)
plt.show()
```

```python
import random
import matplotlib.pyplot as plt

# 1. DADOS DO PROBLEMA

NOME_ITENS = ["Celular", "Notebook", "Livro", "Casaco", "Tênis", "Fone"]
PESOS = [1, 5, 2, 3, 2, 1]
VALORES = [400, 500, 30, 150, 200, 100]

CAPACIDADE_MOCHILA = 8
NUM_GENES = 6
TAM_POPULACAO = 10
NUM_GERACOES = 100
CHANCE_MUTACAO = 0.13

# FUNÇÕES

def criar_individuo():
    return [random.randint(0, 1) for _ in range(NUM_GENES)]

def calcular_fitness(individuo):
    peso = 0
    valor = 0

    for i in range(NUM_GENES):
        if individuo[i] == 1:
            peso += PESOS[i]
            valor += VALORES[i]

    if peso > CAPACIDADE_MOCHILA:
        return 0
    return valor

def selecao_roleta(populacao):
    a = random.choice(populacao)
    b = random.choice(populacao)

    if calcular_fitness(a) > calcular_fitness(b):
        return a
    return b

def crossover(pai1, pai2):
    ponto = NUM_GENES // 2
    filho1 = pai1[:ponto] + pai2[ponto:]
    filho2 = pai2[:ponto] + pai1[ponto:]
    return filho1, filho2

def mutacao(individuo):
    for i in range(NUM_GENES):
        if random.random() < CHANCE_MUTACAO:
            individuo[i] = 1 - individuo[i]

def obter_melhor_individuo(populacao):
    return max(populacao, key=calcular_fitness)

# EXECUÇÃO PRINCIPAL
populacao = [criar_individuo() for _ in range(TAM_POPULACAO)]
lista_melhores_fitness = []

for geracao in range(NUM_GERACOES):

    # 🔥 ELITISMO
    melhor_individuo = obter_melhor_individuo(populacao)
    melhor_fitness = calcular_fitness(melhor_individuo)
    lista_melhores_fitness.append(melhor_fitness)

    nova_populacao = []
    nova_populacao.append(melhor_individuo.copy())  # 👑 elitismo puro

    while len(nova_populacao) < TAM_POPULACAO:
        pai_a = selecao_roleta(populacao)
        pai_b = selecao_roleta(populacao)

        filho_a, filho_b = crossover(pai_a, pai_b)

        mutacao(filho_a)
        mutacao(filho_b)

        nova_populacao.append(filho_a)

        if len(nova_populacao) < TAM_POPULACAO:
            nova_populacao.append(filho_b)

    populacao = nova_populacao
```

```python

# RESULTADOS
print("🧬 Fim da Evolução")
print("Melhor valor encontrado:", lista_melhores_fitness[-1])

plt.plot(lista_melhores_fitness)
plt.title("Evolução do Fitness (com Elitismo)")
plt.xlabel("Geração")
plt.ylabel("Fitness")
plt.grid(True)
plt.show()
```
